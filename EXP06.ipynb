```python
print("### Demonstrating List Methods ###")

# 1. Add (Insert an element using insert())
print("\n--- 1. Add (Insert an element) ---")
my_list_add = [10, 20, 40, 50]
print(f'Original list: {my_list_add}')
my_list_add.insert(2, 30) # Insert 30 at index 2
print(f'List after inserting 30 at index 2: {my_list_add}')

# 2. Append (Add an element to the end using append())
print("\n--- 2. Append (Add an element to the end) ---")
my_list_append = [1, 2, 3]
print(f'Original list: {my_list_append}')
my_list_append.append(4)
print(f'List after appending 4: {my_list_append}')

# 3. Extend (Add multiple elements from an iterable using extend())
print("\n--- 3. Extend (Add multiple elements from an iterable) ---")
my_list_extend = [1, 2, 3]
additional_elements = [4, 5]
print(f'Original list: {my_list_extend}')
print(f'Elements to extend: {additional_elements}')
my_list_extend.extend(additional_elements)
print(f'List after extending with [4, 5]: {my_list_extend}')

# 4. Delete Elements
print("\n--- 4. Delete Elements ---")
print("There are several ways to delete elements from a list:")

# a. remove(): Delete by Value
print("\n  a. remove(): Delete by Value")
my_list_remove = ['apple', 'banana', 'cherry', 'banana']
print(f'Original list: {my_list_remove}')
my_list_remove.remove('banana') # Removes the first occurrence
print(f'List after removing \'banana\': {my_list_remove}')

# b. pop(): Delete by Index (and return the element)
print("\n  b. pop(): Delete by Index (and return the element)")
my_list_pop = [100, 200, 300, 400]
print(f'Original list: {my_list_pop}')
popped_element = my_list_pop.pop(1) # Remove element at index 1
print(f'List after popping element at index 1: {my_list_pop}')
print(f'Popped element: {popped_element}')

# pop() without an index removes and returns the last element
last_element = my_list_pop.pop()
print(f'List after popping last element: {my_list_pop}')
print(f'Last element popped: {last_element}')

# c. del statement: Delete by Index or Slice
print("\n  c. del statement: Delete by Index or Slice")
my_list_del_index = ['a', 'b', 'c', 'd']
print(f'Original list: {my_list_del_index}')
del my_list_del_index[1] # Delete element at index 1
print(f'List after deleting element at index 1: {my_list_del_index}')

# Using del to delete a slice of elements
my_list_del_slice = [1, 2, 3, 4, 5, 6]
print(f'Original list: {my_list_del_slice}')
del my_list_del_slice[1:4] # Delete elements from index 1 up to (but not including) 4
print(f'List after deleting slice [1:4]: {my_list_del_slice}')
```
