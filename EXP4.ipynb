```python
print('1. Nested List:')
nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(f'  {nested_list}')
print(f'  Accessing element [0][1]: {nested_list[0][1]}')

print('\n2. Length of a List:')
my_list = [10, 20, 30, 40, 50]
list_length = len(my_list)
print(f'  Original list: {my_list}')
print(f'  Length of the list: {list_length}')

print('\n3. List Concatenation:')
list1 = [1, 2, 3]
list2 = [4, 5, 6]
concatenated_list = list1 + list2
print(f'  List 1: {list1}')
print(f'  List 2: {list2}')
print(f'  Concatenated list: {concatenated_list}')

print('\n4. Membership Check:')
print(f'  Checking if 3 is in {list1}: {3 in list1}')
print(f'  Checking if 7 is in {list1}: {7 in list1}')

print('\n5. Iteration through a List:')
print(f'  Iterating through {my_list}:')
for item in my_list:
    print(f'    Item: {item}')

print('\n6. List Indexing:')
print(f'  Accessing element at index 0 in {my_list}: {my_list[0]}')
print(f'  Accessing element at index -1 in {my_list}: {my_list[-1]}')

print('\n7. List Slicing:')
print(f'  Slicing from index 1 to 3 (exclusive) in {my_list}: {my_list[1:4]}')
print(f'  Slicing from the beginning to index 3 (exclusive): {my_list[:3]}')
print(f'  Slicing from index 2 to the end: {my_list[2:]}')
print(f'  Slicing with a step of 2: {my_list[::2]}')
```
