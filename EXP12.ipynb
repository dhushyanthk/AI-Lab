```python
from collections import deque

def water_jug_problem(jug1_capacity, jug2_capacity, target):
    """
    Solves the Water Jug Problem using Breadth-First Search (BFS).

    Args:
        jug1_capacity (int): The capacity of the first jug.
        jug2_capacity (int): The capacity of the second jug.
        target (int): The target amount of water to measure.

    Returns:
        list: A list of states (jug1, jug2) representing the solution path,
              or None if no solution is found.
    """
    # Use a deque for BFS queue
    queue = deque()
    # Store visited states to avoid cycles and redundant computations
    visited = set()
    # Store the path to reconstruct the solution
    paths = {}

    # Initial state: both jugs are empty
    initial_state = (0, 0)
    queue.append(initial_state)
    visited.add(initial_state)
    paths[initial_state] = [initial_state] # Path to initial state is just itself

    while queue:
        current_jug1, current_jug2 = queue.popleft()
        current_path = paths[(current_jug1, current_jug2)]

        # Check if the target is reached
        if current_jug1 == target or current_jug2 == target:
            return current_path

        # Generate possible next states
        next_states = []

        # 1. Fill Jug1
        next_states.append((jug1_capacity, current_jug2))
        # 2. Fill Jug2
        next_states.append((current_jug1, jug2_capacity))
        # 3. Empty Jug1
        next_states.append((0, current_jug2))
        # 4. Empty Jug2
        next_states.append((current_jug1, 0))
        # 5. Pour Jug1 to Jug2
        amount_to_pour = min(current_jug1, jug2_capacity - current_jug2)
        next_states.append((current_jug1 - amount_to_pour, current_jug2 + amount_to_pour))
        # 6. Pour Jug2 to Jug1
        amount_to_pour = min(current_jug2, jug1_capacity - current_jug1)
        next_states.append((current_jug1 + amount_to_pour, current_jug2 - amount_to_pour))

        for next_state in next_states:
            if next_state not in visited:
                visited.add(next_state)
                queue.append(next_state)
                # Extend the path to this new state
                paths[next_state] = current_path + [next_state]

    return None # No solution found

# --- Example Usage ---
# Define jug capacities and target amount
jug1_capacity = 4  # Example: 4-gallon jug
jug2_capacity = 3  # Example: 3-gallon jug
target_amount = 2  # Example: Target 2 gallons

solution_path = water_jug_problem(jug1_capacity, jug2_capacity, target_amount)

if solution_path:
    print(f"Solution found for jugs of capacity {jug1_capacity} and {jug2_capacity}, to get {target_amount} gallons:")
    for i, (j1, j2) in enumerate(solution_path):
        print(f"Step {i}: Jug1 = {j1} gallons, Jug2 = {j2} gallons")
else:
    print(f"No solution found for jugs of capacity {jug1_capacity} and {jug2_capacity}, to get {target_amount} gallons.")

# Another example
jug1_capacity = 5
jug2_capacity = 3
target_amount = 4
solution_path = water_jug_problem(jug1_capacity, jug2_capacity, target_amount)

if solution_path:
    print(f"\nSolution found for jugs of capacity {jug1_capacity} and {jug2_capacity}, to get {target_amount} gallons:")
    for i, (j1, j2) in enumerate(solution_path):
        print(f"Step {i}: Jug1 = {j1} gallons, Jug2 = {j2} gallons")
else:
    print(f"\nNo solution found for jugs of capacity {jug1_capacity} and {jug2_capacity}, to get {target_amount} gallons.")
```
