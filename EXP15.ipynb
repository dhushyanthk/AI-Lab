## Vacuum Cleaner Problem

This program simulates a simple vacuum cleaner agent in a rectangular room. The agent's goal is to clean all dirty cells in the room.

```python
import random
import time

class VacuumCleaner:
    def __init__(self, room_size=(5, 5), initial_position=(0, 0)):
        """
        Initializes the vacuum cleaner and the room.
        room_size: A tuple (rows, cols) defining the dimensions of the room.
        initial_position: A tuple (row, col) for the vacuum cleaner's starting position.
        """
        self.rows, self.cols = room_size
        self.room = [['C' for _ in range(self.cols)] for _ in range(self.rows)] # 'C' for Clean
        self.current_pos = initial_position
        self.room[self.current_pos[0]][self.current_pos[1]] = 'V' # 'V' for Vacuum Cleaner
        self.moves = 0
        self.cleaned_cells = 0

    def add_dirt(self, num_dirt_patches=5):
        """
        Randomly adds dirt ('D') to the room.
        """
        dirty_cells_added = 0
        while dirty_cells_added < num_dirt_patches:
            r, c = random.randint(0, self.rows - 1), random.randint(0, self.cols - 1)
            if self.room[r][c] == 'C':
                self.room[r][c] = 'D' # 'D' for Dirty
                dirty_cells_added += 1
        print(f"Added {dirty_cells_added} dirt patches.")

    def display_room(self):
        """
        Prints the current state of the room.
        """
        print("\n--- Room State ---")
        for r in range(self.rows):
            row_str = []
            for c in range(self.cols):
                if (r, c) == self.current_pos:
                    row_str.append('V') # Vacuum cleaner's current position
                else:
                    row_str.append(self.room[r][c])
            print(' '.join(row_str))
        print(f"Current Position: {self.current_pos}, Moves: {self.moves}, Cleaned: {self.cleaned_cells}")
        print("------------------")

    def is_dirty(self, r, c):
        """
        Checks if a cell is dirty.
        """
        return self.room[r][c] == 'D'

    def clean_cell(self):
        """
        Cleans the current cell if it's dirty.
        """
        r, c = self.current_pos
        if self.room[r][c] == 'D':
            self.room[r][c] = 'C'
            self.cleaned_cells += 1
            print(f"Cleaned cell at {self.current_pos}")
            return True
        return False

    def move(self):
        """
        Moves the vacuum cleaner to an adjacent cell (randomly for simplicity).
        It prefers to move towards dirty cells if any are adjacent.
        If no dirty cells adjacent, it moves randomly.
        """
        r, c = self.current_pos
        possible_moves = []
        # Check adjacent cells for dirt and valid moves
        neighbors = [
            (r-1, c), (r+1, c), (r, c-1), (r, c+1) # Up, Down, Left, Right
        ]

        dirty_neighbors = []
        clean_neighbors = []

        for nr, nc in neighbors:
            if 0 <= nr < self.rows and 0 <= nc < self.cols:
                if self.is_dirty(nr, nc):
                    dirty_neighbors.append((nr, nc))
                else:
                    clean_neighbors.append((nr, nc))

        if dirty_neighbors:
            next_pos = random.choice(dirty_neighbors)
        elif clean_neighbors:
            next_pos = random.choice(clean_neighbors)
        else: # Stuck, no moves possible (shouldn't happen in typical grid)
            return False

        self.current_pos = next_pos
        self.moves += 1
        return True

    def run(self, max_moves=100, delay=0.5):
        """
        Runs the vacuum cleaner simulation until all dirt is cleaned or max_moves is reached.
        """
        initial_dirty_cells = sum(row.count('D') for row in self.room)
        if initial_dirty_cells == 0:
            print("Room is already clean!")
            self.display_room()
            return

        print("Starting vacuum cleaner simulation...")
        self.display_room()
        time.sleep(delay)

        while self.cleaned_cells < initial_dirty_cells and self.moves < max_moves:
            self.clean_cell()
            if self.cleaned_cells == initial_dirty_cells:
                break # All dirty cells cleaned
            self.move()
            self.display_room()
            time.sleep(delay)

        print("\n--- Simulation Ended ---")
        if self.cleaned_cells == initial_dirty_cells:
            print(f"All {initial_dirty_cells} dirty cells cleaned in {self.moves} moves!")
        else:
            print(f"Max moves ({max_moves}) reached. Cleaned {self.cleaned_cells} of {initial_dirty_cells} dirty cells.")
        self.display_room()
```

```python
# Create a 5x5 room with the vacuum starting at (0,0)
vacuum = VacuumCleaner(room_size=(5, 5), initial_position=(0, 0))

# Add 7 patches of dirt randomly
vacuum.add_dirt(num_dirt_patches=7)

# Run the simulation for a maximum of 50 moves with a 0.2 second delay between steps
vacuum.run(max_moves=50, delay=0.2)
```
