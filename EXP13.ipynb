```python
import itertools

def solve_cryptarithmetic(puzzle):
    words = puzzle.replace(' ', '').replace('+', ' ').replace('=', ' ').split()
    unique_chars = sorted(list(set(''.join(words))))

    # The first letter of each word cannot be zero
    leading_chars = {word[0] for word in words}

    if len(unique_chars) > 10:
        return "Error: More than 10 unique characters, impossible to solve."

    for perm in itertools.permutations(range(10), len(unique_chars)):
        char_to_digit = dict(zip(unique_chars, perm))

        # Check if leading characters are not zero
        if any(char_to_digit[char] == 0 for char in leading_chars):
            continue

        # Convert words to numbers and evaluate the expression
        nums = []
        for word in words:
            num = 0
            for char in word:
                num = num * 10 + char_to_digit[char]
            nums.append(num)

        # Split into operands and result based on the puzzle structure
        # This assumes the format is OPERAND1 + OPERAND2 = RESULT
        # More complex puzzles would require more robust parsing
        operands = nums[:-1]
        result = nums[-1]

        if sum(operands) == result:
            # Return the mapping if a solution is found
            solution = {char: digit for char, digit in char_to_digit.items()}
            return solution

    return "No solution found."
```

```python
# Classic SEND + MORE = MONEY puzzle
puzzle1 = "SEND + MORE = MONEY"
solution1 = solve_cryptarithmetic(puzzle1)

if isinstance(solution1, dict):
    print(f"Solution for '{puzzle1}':")
    for char, digit in solution1.items():
        print(f"{char} = {digit}")

    # Verify the solution
    send = solution1['S']*1000 + solution1['E']*100 + solution1['N']*10 + solution1['D']
    more = solution1['M']*1000 + solution1['O']*100 + solution1['R']*10 + solution1['E']
    money = solution1['M']*10000 + solution1['O']*1000 + solution1['N']*100 + solution1['E']*10 + solution1['Y']
    print(f"Verification: {send} + {more} = {money} (Expected: {send + more})")

elif isinstance(solution1, str):
    print(solution1)

print("\n---\n")

# Another example
puzzle2 = "A + B = C"
solution2 = solve_cryptarithmetic(puzzle2)

if isinstance(solution2, dict):
    print(f"Solution for '{puzzle2}':")
    for char, digit in solution2.items():
        print(f"{char} = {digit}")
elif isinstance(solution2, str):
    print(solution2)
```
