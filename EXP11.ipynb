## 8-Queens Problem Solver

The 8-Queens problem is a classic combinatorial problem where the goal is to place eight chess queens on an 8×8 chessboard such that no two queens threaten each other. This means no two queens can share the same row, column, or diagonal.

The solution typically involves a **backtracking algorithm**:

1.  **Start with the first column (or row).**
2.  **Place a queen in the first available safe square** in the current column.
3.  **Recursively try to place a queen in the next column.**
4.  If a queen cannot be placed safely in the current column, **backtrack** to the previous column and try a different position for the queen there.
5.  If all queens are placed successfully, a solution is found.

```python
def solve_n_queens(n):
    """
    Solves the N-Queens problem and returns all possible solutions.
    A solution is represented as a list of column indices for each row.
    """
    board = [-1] * n  # board[i] = column of queen in row i
    solutions = []

    def is_safe(row, col):
        # Check if placing a queen at (row, col) is safe
        for prev_row in range(row):
            # Check column conflict
            if board[prev_row] == col:
                return False
            # Check diagonal conflict
            if abs(board[prev_row] - col) == abs(prev_row - row):
                return False
        return True

    def backtrack(row):
        if row == n:
            # All queens placed, add current configuration to solutions
            solutions.append(list(board))
            return

        for col in range(n):
            if is_safe(row, col):
                board[row] = col
                backtrack(row + 1)
                board[row] = -1  # Backtrack: remove queen for next attempt

    backtrack(0)  # Start the backtracking process from row 0
    return solutions

def print_solution(solution, n):
    """
    Prints a single N-Queens solution graphically.
    """
    print("--- Solution ---")
    for row in range(n):
        line = []
        for col in range(n):
            if solution[row] == col:
                line.append('Q') # Queen at this position
            else:
                line.append('.') # Empty square
        print(" ".join(line))
    print("----------------")
```

```python
# Solve the 8-Queens problem
n_queens = 8
all_solutions = solve_n_queens(n_queens)

print(f"Found {len(all_solutions)} solutions for the {n_queens}-Queens problem.")

# Print the first 5 solutions (if available)
for i, solution in enumerate(all_solutions[:5]):
    print(f"\nSolution {i+1}:")
    print_solution(solution, n_queens)

# You can change `n_queens` to a smaller number like 4 for a quicker demonstration:
# print("\n\nSolving 4-Queens problem:")
# four_queens_solutions = solve_n_queens(4)
# for i, solution in enumerate(four_queens_solutions):
#     print(f"\nSolution {i+1}:")
#     print_solution(solution, 4)
```
