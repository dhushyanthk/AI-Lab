```python
class State:
    def __init__(self, missionaries, cannibals, boat):
        self.missionaries = missionaries
        self.cannibals = cannibals
        self.boat = boat  # 0 for left bank, 1 for right bank

    def is_valid(self):
        if self.missionaries < 0 or self.missionaries > 3:
            return False
        if self.cannibals < 0 or self.cannibals > 3:
            return False

        # Check if cannibals outnumber missionaries on either bank
        if 0 < self.missionaries < self.cannibals:
            return False

        # Check the right bank (total - current on left)
        rm = 3 - self.missionaries
        rc = 3 - self.cannibals
        if 0 < rm < rc:
            return False

        return True

    def is_goal(self):
        return self.missionaries == 0 and self.cannibals == 0 and self.boat == 1

    def __eq__(self, other):
        return (self.missionaries == other.missionaries and
                self.cannibals == other.cannibals and
                self.boat == other.boat)

    def __hash__(self):
        return hash((self.missionaries, self.cannibals, self.boat))

    def __str__(self):
        return f"M: {self.missionaries}, C: {self.cannibals}, Boat: {'Left' if self.boat == 0 else 'Right'}"

def solve_missionaries_cannibals():
    start_state = State(3, 3, 0)
    # Each element in the queue is a tuple: (current_state, path_to_current_state)
    queue = [(start_state, [start_state])]
    visited = {start_state}

    while queue:
        current_state, path = queue.pop(0)

        if current_state.is_goal():
            return path

        # Possible moves (missionaries, cannibals)
        # Max 2 people in the boat
        moves = [
            (1, 0),  # 1 missionary
            (2, 0),  # 2 missionaries
            (0, 1),  # 1 cannibal
            (0, 2),  # 2 cannibals
            (1, 1)   # 1 missionary, 1 cannibal
        ]

        for dm, dc in moves:
            if current_state.boat == 0:  # Boat on the left bank (moving to right)
                new_m = current_state.missionaries - dm
                new_c = current_state.cannibals - dc
                new_boat = 1
            else:  # Boat on the right bank (moving to left)
                new_m = current_state.missionaries + dm
                new_c = current_state.cannibals + dc
                new_boat = 0

            next_state = State(new_m, new_c, new_boat)

            if next_state.is_valid() and next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [next_state]))

    return None # No solution found

solution = solve_missionaries_cannibals()

if solution:
    print("Solution found:")
    for i, state in enumerate(solution):
        print(f"Step {i}: {state}")
else:
    print("No solution exists.")
```
