```python
import string

def remove_punctuation(text):
    """Removes punctuation from a string."""
    # Create a translation table to map punctuation characters to None
    translator = str.maketrans('', '', string.punctuation)
    # Use the translate method to remove punctuation
    return text.translate(translator)

# Example usage:
example_string = "Hello, world! This is a test string with some punctuation."
cleaned_string = remove_punctuation(example_string)
print(f"Original string: {example_string}")
print(f"Cleaned string: {cleaned_string}")
```
