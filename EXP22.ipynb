````python
# Define the Minimax algorithm
def minimax(board, depth, is_maximizing_player):
    # Base cases: Check for terminal states (win/loss/draw) or depth limit
    score = evaluate(board)
    if score == 10:  # Maximizing player wins
        return score - depth
    if score == -10: # Minimizing player wins
        return score + depth
    if not any(' ' in row for row in board): # No empty cells left, it's a draw
        return 0

    if is_maximizing_player:
        best_score = -float('inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    board[i][j] = 'X'
                    best_score = max(best_score, minimax(board, depth + 1, not is_maximizing_player))
                    board[i][j] = ' '
        return best_score
    else:
        best_score = float('inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    board[i][j] = 'O'
                    best_score = min(best_score, minimax(board, depth + 1, not is_maximizing_player))
                    board[i][j] = ' '
        return best_score

def evaluate(board):
    # Check rows for win
    for row in range(3):
        if board[row][0] == board[row][1] and board[row][1] == board[row][2]:
            if board[row][0] == 'X':
                return 10
            elif board[row][0] == 'O':
                return -10

    # Check columns for win
    for col in range(3):
        if board[0][col] == board[1][col] and board[1][col] == board[2][col]:
            if board[0][col] == 'X':
                return 10
            elif board[0][col] == 'O':
                return -10

    # Check diagonals for win
    if board[0][0] == board[1][1] and board[1][1] == board[2][2]:
        if board[0][0] == 'X':
            return 10
        elif board[0][0] == 'O':
            return -10

    if board[0][2] == board[1][1] and board[1][1] == board[2][0]:
        if board[0][2] == 'X':
            return 10
        elif board[0][2] == 'O':
            return -10

    return 0 # No winner yet

def find_best_move(board):
    best_val = -float('inf')
    best_move = (-1, -1)

    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                board[i][j] = 'X'
                move_val = minimax(board, 0, False)
                board[i][j] = ' '

                if move_val > best_val:
                    best_move = (i, j)
                    best_val = move_val
    return best_move

def print_board(board):
    for row in board:
        print(" | ".join(row))
        print("-----------")
````

````python
# Example usage (simple Tic-Tac-Toe game)
board = [
    [' ', ' ', ' '],
    [' ', ' ', ' '],
    [' ', ' ', ' ']
]

print("Initial Board:")
print_board(board)

# Let's make a few moves and see Minimax in action

# Player X (AI) makes a move
best_move = find_best_move(board)
board[best_move[0]][best_move[1]] = 'X'
print("\nAI (X) move:")
print_board(board)

# Player O (Human or another AI) makes a move
board[0][0] = 'O'
print("\nPlayer O move (e.g., top-left):")
print_board(board)

# Player X (AI) makes another move
best_move = find_best_move(board)
board[best_move[0]][best_move[1]] = 'X'
print("\nAI (X) move:")
print_board(board)

# Player O makes another move
board[1][0] = 'O'
print("\nPlayer O move (e.g., middle-left):")
print_board(board)

# Player X (AI) makes another move
best_move = find_best_move(board)
board[best_move[0]][best_move[1]] = 'X'
print("\nAI (X) move:")
print_board(board)

# You can continue the game by making moves for 'O' and then calling find_best_move for 'X'
# and checking the evaluate function for win/draw conditions.
````
