```python
from collections import deque

def bfs(graph, start_node):
    """
    Implements the Breadth-First Search algorithm.

    Args:
        graph (dict): An adjacency list representation of the graph.
                      e.g., {'A': ['B', 'C'], 'B': ['D', 'E'], ...}
        start_node: The node from which to start the BFS.

    Returns:
        list: A list of nodes in the order they were visited by BFS.
    """
    visited = set()
    queue = deque([start_node])
    bfs_order = []

    visited.add(start_node)

    while queue:
        current_node = queue.popleft()
        bfs_order.append(current_node)

        for neighbor in graph.get(current_node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return bfs_order

# Example Usage:
# Define a sample graph using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

start_node = 'A'
print(f"BFS starting from node {start_node}: {bfs(graph, start_node)}")

start_node_2 = 'C'
print(f"BFS starting from node {start_node_2}: {bfs(graph, start_node_2)}")
```
