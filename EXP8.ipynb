```python
import calendar

def generate_calendar(year, month):
  """Generates and prints a calendar for the given month and year."""
  # Create a TextCalendar instance
  cal = calendar.TextCalendar(calendar.SUNDAY) # Start week on Sunday

  # Format the calendar for the given month and year
  # pragma: allowlist nextline
  print(cal.formatmonth(year, month, w=2, l=1)) # w=width of date column, l=lines per week

# Example usage:
# Let's generate a calendar for October 2023
print("Calendar for October 2023:")
generate_calendar(2023, 10)

# You can try another example, e.g., current month and year
import datetime
current_year = datetime.date.today().year
current_month = datetime.date.today().month

print(f"\nCalendar for {datetime.date.today().strftime('%B %Y')}:")
generate_calendar(current_year, current_month)
```
