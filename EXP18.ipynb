```python
import math

def solve_tsp_dp(distances):
    """
    Solves the Traveling Salesman Problem using dynamic programming.

    Args:
        distances (list of lists): A square matrix representing the distances
                                   between cities. distances[i][j] is the distance
                                   from city i to city j.

    Returns:
        tuple: A tuple containing the minimum cost and the optimal path.
    """
    n = len(distances)

    # dp[mask][i] stores the minimum cost to visit cities in 'mask'
    # ending at city 'i'.
    dp = [[float('inf')] * n for _ in range(1 << n)]

    # parent[mask][i] stores the predecessor city of 'i' in the optimal path
    # for the given mask.
    parent = [[-1] * n for _ in range(1 << n)]

    # Base case: Starting from city 0, visiting only city 0.
    dp[1 << 0][0] = 0

    # Iterate over all possible masks (subsets of cities)
    for mask in range(1, 1 << n):
        # Iterate over all cities 'i' that can be the last city in the path represented by 'mask'
        for i in range(n):
            # If city 'i' is in the current mask
            if (mask >> i) & 1:
                # Iterate over all possible previous cities 'j'
                for j in range(n):
                    if i == j: # Cannot be the same city
                        continue
                    # If city 'j' is in the mask, and 'i' is not 'j'
                    if (mask >> j) & 1: # Check if city j is in the mask
                        # Calculate the cost if we come from city 'j' to city 'i'
                        if dp[mask ^ (1 << i)][j] + distances[j][i] < dp[mask][i]:
                            dp[mask][i] = dp[mask ^ (1 << i)][j] + distances[j][i]
                            parent[mask][i] = j

    # Find the minimum cost to complete the cycle back to city 0 from any last city
    min_cost = float('inf')
    last_city = -1
    full_mask = (1 << n) - 1

    for i in range(n):
        if dp[full_mask][i] + distances[i][0] < min_cost:
            min_cost = dp[full_mask][i] + distances[i][0]
            last_city = i

    # Reconstruct the path
    path = []
    current_mask = full_mask
    current_city = last_city

    while current_city != -1:
        path.append(current_city)
        prev_city = parent[current_mask][current_city]
        current_mask ^= (1 << current_city) # Remove current_city from mask
        current_city = prev_city

    path.append(0) # Add the starting city to complete the cycle
    path.reverse()

    return min_cost, path
```

```python
# Example Usage:
# Define a distance matrix for 4 cities (A, B, C, D)
# distances[i][j] is the distance from city i to city j
# City 0: A, City 1: B, City 2: C, City 3: D

distances = [
    [0, 10, 15, 20],  # A to A, B, C, D
    [10, 0, 35, 25],  # B to A, B, C, D
    [15, 35, 0, 30],  # C to A, B, C, D
    [20, 25, 30, 0]   # D to A, B, C, D
]

min_cost, path = solve_tsp_dp(distances)

print(f"Minimum cost for TSP: {min_cost}")
print(f"Optimal path: {path}")

# Another example with different distances
distances_2 = [
    [0, 29, 20, 21],
    [29, 0, 15, 17],
    [20, 15, 0, 28],
    [21, 17, 28, 0]
]

min_cost_2, path_2 = solve_tsp_dp(distances_2)

print(f"\nMinimum cost for TSP (example 2): {min_cost_2}")
print(f"Optimal path (example 2): {path_2}")
```
